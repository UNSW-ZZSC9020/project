```python
import pandas as pd
import plotly.express as px
```

```python
gantt_data_df = pd.read_csv('gantt_chart_data.csv')
gantt_data_df['Start'] = pd.to_datetime(gantt_data_df['Start'], dayfirst=True)
gantt_data_df['Finish'] = pd.to_datetime(gantt_data_df['Finish'], dayfirst=True)
```

```python
week_boundaries_df = gantt_data_df.groupby('Week').agg({'Start': 'min', 'Finish': 'max'}).reset_index()
week_boundaries = [(i['Start'], i['Finish']) for i in week_boundaries_df.to_dict('records')]
```

```python
WEEK_COLOURS = {
    "Week 1": "#E64F4F",
    "Week 2": "#F0903C",
    "Week 3": "#DFD938",
    "Week 4": "#24A61D",
    "Week 5": "#2C38D8",
    "Week 6": "#B059E2",
    "Week 7": "#6638FA",
}
fig = px.timeline(
    gantt_data_df,
    x_start="Start",
    x_end="Finish",
    y="Task",
    color="Week",
    color_discrete_map=WEEK_COLOURS,
    title="Group A Project Timeline",
)
fig.update_layout(
    height=720,
    width=1500,
    plot_bgcolor="white",
    legend_title=None,
    font={'size': 15},
    title_font_size=26,
    yaxis_title="",
)

for i, (x0, x1) in enumerate(week_boundaries):
    fig.add_vrect(
        x0=x0, x1=x1,
        fillcolor="#F5F5F5" if i % 2 == 0 else "white",
        layer="below",
        line_width=0,
    )
fig.update_yaxes(autorange="reversed")
fig.show()
```
