```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from ModelFiles.LSTM.LSTMUtils import *
from ModelFiles.ModelConfigs import LSTMConfig, SEEDS
from ModelFiles.ModelEnums import LSTMModelType
from ModelFiles.ModelPlots import *

use_log_target = True
EVAL_STEP_SIZE = 48
NUM_EPOCHS = 100
PATIENCE = 10
DEBUG = False
HORIZON_LOOKBACK = [(48, 168), (336, 336), (720, 720)]
for horizon, lookback in HORIZON_LOOKBACK:
    for seed in SEEDS:
        if seed == SEEDS[0]: # only save results for the first seed to save space
            save_prediction_results = True
        else:
            save_prediction_results = False
        biLSTM_config = LSTMConfig(
            task_id= f"biLSTM_best_model_h{horizon}_lb{lookback}_seed{seed}",
            model_type= LSTMModelType.BILSTM,
            hidden_size= 128,
            num_layers= 2,
            dropout= 0.2,
            learning_rate= 0.00001,
            batch_size= 64,
            training_epochs= NUM_EPOCHS,
            patience= PATIENCE,
            use_mlp_head= True,
            mlp_hidden_size= 64,
            target_col= "TOTALDEMAND" if not use_log_target else "LOG_TOTALDEMAND",
            used_log_target= use_log_target,
            target_lags= [48, 336],
            target_mas= [],
            feature_cols= ["TEMPERATURE"],
            feature_lag_cols= [],
            lookback_window= lookback,
            forecast_horizon= horizon,
            forecast_last_step_only=False,
            num_attention_heads= 4,
            weight_decay= 1e-4,
            scale=True,
            seed=seed,
            save_training_log= False,
            save_test_results= save_prediction_results,
            eval_step_size=EVAL_STEP_SIZE,
            debug= DEBUG,
            save_model=False
        )
        runs_df_artifact = run_experiment(config=biLSTM_config, experiment_name=biLSTM_config.task_id, specific_output_dir=SAVE_PATH)
        print("=" * 200)
        print("\n")
```
